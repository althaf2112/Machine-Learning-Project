# Insights

Based on exploratory data analysis, feature engineering, and model evaluation, several important insights were discovered from the customer churn dataset.

1. **Customer Activity Influences Churn**  
Customers with lower activity rates (fewer transactions relative to account age) tend to have a higher likelihood of churn. Active customers are more engaged with the bank's services.

2. **Credit Utilization Patterns**  
Customers with higher credit usage ratios show a higher tendency to churn. High utilization may indicate financial stress or dissatisfaction with credit limits.

3. **Transaction Behavior**  
Customers with higher transaction frequency generally show stronger engagement with the bank, making them less likely to leave.

4. **Customer Tenure Matters**  
Customers who have been with the bank for a longer period tend to be more loyal and have lower churn rates.

5. **Spending Patterns**  
Average transaction value does not strongly differentiate churned customers from existing customers, suggesting that spending size alone is not a strong churn indicator.

6. **Model Performance**  
The Logistic Regression model achieved approximately 83.9% accuracy with a recall score of 1.0, indicating that the model successfully detected all churned customers.

7. **False Churn Predictions**  
The confusion matrix shows that some loyal customers were incorrectly predicted as churned, which indicates the model tends to overestimate churn risk.

# Conclusion

This project aimed to analyze customer behavior and build a predictive model to identify customers who are likely to churn.

Through data cleaning, exploratory data analysis, and feature engineering, several important behavioral patterns were identified. Features such as transaction activity, credit utilization, and customer tenure were found to provide meaningful insights into churn behavior.

A Logistic Regression model was trained to predict customer churn and achieved strong performance with an accuracy of approximately 83.9%. The model was particularly effective at identifying churned customers, achieving a recall score of 1.0.

These findings demonstrate that predictive analytics can help financial institutions proactively identify at-risk customers and implement strategies to improve customer retention.

# Limitations
Despite the useful insights obtained, this analysis has several limitations.

The dataset does not include external factors such as customer satisfaction, service quality, or competitor offers that may influence churn behavior.

Additionally, Logistic Regression assumes a linear relationship between features and the target variable, which may not fully capture complex customer behaviors.

# Possible Improvements
Future improvements could enhance the predictive performance and insights of the model.

More advanced machine learning models such as Random Forest, Gradient Boosting, or XGBoost could be explored to capture complex patterns.

Additional features such as customer engagement metrics, service usage history, or feedback scores could provide deeper insights into churn behavior.

Hyperparameter tuning and cross-validation could also be applied to improve model performance.

# Recommended Actions
Based on the analysis and predictive modeling results, the following actions are recommended to reduce customer churn:

Target High-Risk Customers Customers predicted to churn should be contacted proactively with retention offers such as loyalty rewards, reduced fees, or personalized financial advice.

Improve Customer Engagement Encourage customers to increase transaction activity through promotions, cashback offers, or reward programs.

Monitor High Credit Utilization Customers Customers with high credit usage should be monitored carefully and offered financial planning support or credit limit adjustments.

Strengthen Customer Relationships Customers with longer tenure should be provided with exclusive benefits to maintain loyalty and prevent potential churn.

Implement Early Warning Systems Use the predictive churn model as an early warning system to identify at-risk customers before they leave.

Customer Feedback Programs Collect feedback from customers predicted to churn to understand dissatisfaction and improve banking services.

Continuous Model Improvement Regularly update and retrain the churn prediction model with new data to improve accuracy and adapt to changing customer behavior.
