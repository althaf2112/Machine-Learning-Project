# Data Cleaning and Preprocessing

Data cleaning is an important step in data analysis and machine learning. Raw datasets often contain missing values, duplicate records, inconsistent formats, or categorical variables that cannot be directly used in machine learning models.

In this section, we will clean the dataset and prepare it for modeling by:

- Handling missing values
- Removing duplicate records
- Fixing inconsistent values
- Converting data types
- Encoding categorical variables

```python
import pandas as pd
import numpy as np

df = pd.read_csv("day1_ml_output.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['Education_Level'].unique()
```

```python
df['Education_Level'] = df['Education_Level'].replace('Unknown','Other')
```

## Fixing Inconsistent Values

Some categorical variables contained inconsistent values such as `"Unknown"`. These values were replaced with a more general category to maintain consistency.

```python
df.dtypes
```

## Label Encoding (Binary Categories)

Binary categorical variables such as `Gender` and `Attrition_Flag` were converted into numerical values using Label Encoding. This allows machine learning algorithms to process the data.

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Gender'] = le.fit_transform(df['Gender'])
df['Attrition_Flag'] = le.fit_transform(df['Attrition_Flag'])
```

## One-Hot Encoding (Multiple Categories)

Categorical variables with multiple categories were transformed using One-Hot Encoding. This method creates separate binary columns for each category and prevents models from interpreting categorical values as ordinal relationships.

```python
df = pd.get_dummies(df, columns=[
    'Education_Level',
    'Marital_Status',
    'Income_Category',
    'Card_Category'
])
```

```python
bool_cols = df.select_dtypes(include='bool').columns
df[bool_cols] = df[bool_cols].astype(int)
```

After preprocessing, the dataset is now clean and ready for feature engineering and modeling.
